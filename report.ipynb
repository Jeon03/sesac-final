```python
import requests
from bs4 import BeautifulSoup

# 1. 대상 URL
url = "https://www.zionmarketresearch.com/report/us-skincare-market"
headers = {"User-Agent": "Mozilla/5.0"}

try:
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    soup = BeautifulSoup(response.text, 'html.parser')

    # 2. 원문 컨테이너 추출 (보여주신 구조 기준)
    content_div = soup.find('div', id='Descriptionss')

    if content_div:
        # 모든 텍스트를 문단별로 줄바꿈하여 추출
        full_text = content_div.get_text(separator="\n\n", strip=True)
        
        # 3. 파일로 저장 (나중에 복사해서 LLM에 붙여넣기용)
        with open("us_skincare_full_text.txt", "w", encoding="utf-8") as f:
            f.write(full_text)
            
        print("✅ 원문 추출 및 파일 저장 완료!")
        print("-" * 30)
        print(full_text[:300]) # 샘플 확인
    else:
        print("❌ 컨테이너를 찾지 못했습니다.")

except Exception as e:
    print(f"⚠️ 에러: {e}")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time

# 1. 브라우저 및 드라이버 설정
chrome_options = Options()
chrome_options.add_argument("--headless")  # 창 없이 실행
chrome_options.add_argument("--no-sandbox")
chrome_options.add_argument("--disable-dev-shm-usage")
# 실제 브라우저처럼 보이게 유저 에이전트 설정
chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36")

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=chrome_options)

# 2. 대상 URL
url = "https://biz.chosun.com/en/en-retail/2025/03/14/4MAZV5QDYJHPRMHAELGWZOLPU4/"

try:
    print(f"🌐 페이지 접속 중: {url}")
    driver.get(url)
    
    # 3. 본문 요소가 나타날 때까지 최대 10초 대기 (Explicit Wait)
    # itemprop="articleBody" 속성을 가진 section 태그를 기다립니다.
    wait = WebDriverWait(driver, 10)
    article_element = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "section[itemprop='articleBody']")))

    # 자바스크립트 실행 완료를 위해 2초 추가 대기
    time.sleep(2)

    # 4. 불필요한 요소 제거 (이미지 캡션, 광고 영역 등)
    # Selenium의 execute_script를 사용해 브라우저 상에서 직접 삭제합니다.
    driver.execute_script("""
        var figures = document.querySelectorAll('figure');
        figures.forEach(el => el.remove());
        var ads = document.querySelectorAll('.arcad-wrapper, #taboola-mid-article-widget-1x1');
        ads.forEach(el => el.remove());
        var trans_info = document.querySelectorAll('.text--grey-60');
        trans_info.forEach(el => el.remove());
    """)

    # 5. 텍스트 추출 및 정제
    # .text는 렌더링된 텍스트만 가져오므로 매우 깔끔합니다.
    full_text = article_element.text.strip()
    
    if full_text:
        # 6. 파일 저장
        filename = "qoo10_japan_biz_selenium.txt"
        with open(filename, "w", encoding="utf-8") as f:
            f.write(full_text)
            
        print("✅ 원문 추출 및 파일 저장 완료!")
        print("-" * 50)
        print(f"추출된 글자 수: {len(full_text)}자")
        print("--- 본문 내용 샘플 ---")
        print(full_text[:500])
    else:
        print("❌ 본문 내용은 찾았으나 텍스트가 비어있습니다.")

except Exception as e:
    print(f"⚠️ 에러 발생: {e}")

finally:
    driver.quit()
```
